```python
import numpy as np
import pandas as pd

df = pd.read_csv('dataset/dataset_1.csv', index_col=0)
print(df)
print(df.describe())

```

```python
print(df.describe())
```

```python
print(df.info())    
```

```python
set_gen = set(df["Género"].to_list())
set_edu = set(df["Nivel_Educación"].to_list())
set_ciu = set(df["Ciudad"].to_list())
print(set_gen)
print(set_edu)
print(set_ciu)
```

```python
#tratamiento de datos de valores negativos 
df["Edad"] = df["Edad"].apply(lambda x: np.nan if x < 0 else x)
df["Ingresos"] = df["Ingresos"].apply(lambda x: np.nan if x < 0 else x)
df["Hijos"] = df["Hijos"].apply(lambda x: np.nan if x < 0 else x)
# tratamientos de datos faltantes
for column in ["Edad","Ingresos","Hijos"]:
    median_value = df[column].median()
    df[column]= df[column].fillna(median_value)

for column in ["Género","Ciudad"]:
    mode_value  = df[column].mode()[0]
    df[column]= df[column].fillna(mode_value)


# 3 mapeo de datos de todos los datos categoricos a numericos
education_mapping = {
    "Bachelors" : "Bachelor",
    "mastre" : "Master",
    "pHd" : "PhD",
    "no education" : "Ne"
}

df["Nivel_Educación"] = df["Nivel_Educación"].replace(education_mapping)
df["Nivel_Educación"]= df["Nivel_Educación"].fillna("Ne")

df["Edad"] = df["Edad"].astype(int)
df["Hijos"] = df["Hijos"].astype(int)
df["Ingresos"] = df["Ingresos"].astype(float)
df["Altura"] = df["Altura"].astype(float)
```

```python
#casteo de tipo de datos 
print(df)
print(df.info())
print(df.describe())
```
